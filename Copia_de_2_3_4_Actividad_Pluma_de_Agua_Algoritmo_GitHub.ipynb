````python
# ==========================================================
# ACTIVIDAD: MODELAR LA "PLUMA DE AGUA" CON PYTHON
# ==========================================================

# 1. Solicitar al usuario el número de plumas
plumas = float(input("Ingrese el número de plumas: "))

# 2. Definir constantes del problema
CAUDAL_PLUMA = 0.025  # Litros por segundo por pluma
SEGUNDOS_DIA = 24 * 60 * 60  # Segundos en un día
DIAS_MES = 30  # Días en un mes para el cálculo mensual
DIAS_SEMANA = 7 # Días en una semana
LITROS_POR_METRO_CUBICO = 1000 # Conversión de litros a metros cúbicos

# 3. Calcular el caudal total (litros por segundo)
caudal_total = plumas * CAUDAL_PLUMA

# 4. Calcular el consumo diario (litros)
consumo_diario = caudal_total * SEGUNDOS_DIA

# 5. Calcular el consumo semanal (litros)
consumo_semanal = consumo_diario * DIAS_SEMANA

# 6. Calcular el consumo mensual (litros)
consumo_mensual = consumo_diario * DIAS_MES

# 7. Convertir consumo a metros cúbicos
consumo_diario_m3 = consumo_diario / LITROS_POR_METRO_CUBICO
consumo_semanal_m3 = consumo_semanal / LITROS_POR_METRO_CUBICO
consumo_mensual_m3 = consumo_mensual / LITROS_POR_METRO_CUBICO

# 8. Estimación del costo (opcional)
costo_por_m3 = float(input("Ingrese el costo por metro cúbico (ej. 2.5): "))
costo_diario = consumo_diario_m3 * costo_por_m3
costo_semanal = consumo_semanal_m3 * costo_por_m3
costo_mensual = consumo_mensual_m3 * costo_por_m3

# 9. Clasificar el consumo diario
nivel_consumo = ""
if consumo_diario < 1000:
    nivel_consumo = "Bajo"
elif 1000 <= consumo_diario <= 3000:
    nivel_consumo = "Medio"
else:
    nivel_consumo = "Alto"

# 10. Mostrar resultados
print(f"\n--- Análisis de Consumo para {plumas} pluma(s) de agua ---")
print(f"\nCaudal y Consumo en Litros:")
print(f"  Caudal total: {caudal_total:.4f} litros/segundo")
print(f"  Consumo diario estimado: {consumo_diario:.2f} litros")
print(f"  Consumo semanal estimado: {consumo_semanal:.2f} litros")
print(f"  Consumo mensual estimado (suponiendo {DIAS_MES} días): {consumo_mensual:.2f} litros")

print(f"\nConsumo en Metros Cúbicos:")
print(f"  Consumo diario estimado: {consumo_diario_m3:.4f} m³")
print(f"  Consumo semanal estimado: {consumo_semanal_m3:.4f} m³")
print(f"  Consumo mensual estimado: {consumo_mensual_m3:.4f} m³")

print(f"\nEstimación de Costos (con tarifa de {costo_por_m3:.2f} por m³):")
print(f"  Costo diario estimado: {costo_diario:.2f}")
print(f"  Costo semanal estimado: {costo_semanal:.2f}")
print(f"  Costo mensual estimado: {costo_mensual:.2f}")

print(f"\nClasificación del consumo diario: {nivel_consumo}")
print("--------------------------------------------------")
````

```python

plumas = float(input("Ingrese el número de plumas: "))

# 2. Definir constantes del problema
# CAUDAL_PLUMA: Define el caudal de una pluma en litros por segundo, una constante histórica.

SEGUNDOS_DIA = 24 * 60 * 60  # Segundos en un día

DIAS_SEMANA = 7 # Días en una semana

# El consumo diario se calcula multiplicando el caudal total por los segundos que tiene un día.
consumo_diario = caudal_total * SEGUNDOS_DIA

# 5. Calcular el consumo semanal (litros)
# El consumo semanal se obtiene multiplicando el consumo diario por el número de días en una semana.
consumo_semanal = consumo_diario * DIAS_SEMANA

# 6. Calcular el consumo mensual (litros)
# El consumo mensual se calcula multiplicando el consumo diario por el número de días en un mes.
consumo_mensual = consumo_diario * DIAS_MES

# 7. Convertir consumo a metros cúbicos
# Se convierten los consumos diarios, semanales y mensuales de litros a metros cúbicos
# dividiéndolos por el factor de conversión LITROS_POR_METRO_CUBICO.
consumo_diario_m3 = consumo_diario / LITROS_POR_METRO_CUBICO
consumo_semanal_m3 = consumo_semanal / LITROS_POR_METRO_CUBICO
consumo_mensual_m3 = consumo_mensual / LITROS_POR_METRO_CUBICO

# 8. Estimación del costo (opcional)
# Se solicita al usuario el costo por metro cúbico para calcular los costos asociados al consumo.
costo_por_m3 = float(input("Ingrese el costo por metro cúbico (ej. 2.5): "))
# Se calculan los costos diarios, semanales y mensuales multiplicando el consumo en m3 por el costo por m3.

costo_semanal = consumo_semanal_m3 * costo_por_m3

    nivel_consumo = "Bajo"

elif 1000 <= consumo_diario <= 3000:
    nivel_consumo = "Medio"
# "Alto" si es mayor a 3000 litros.
else:
    nivel_consumo = "Alto"

# 10. Mostrar resultados
# Se imprimen todos los resultados de manera formateada y clara usando f-strings.
print(f"\n--- Análisis de Consumo para {plumas} pluma(s) de agua ---")

print(f"\nCaudal y Consumo en Litros:")
print(f"  Caudal total: {caudal_total:.4f} litros/segundo") # Formateado a 4 decimales.

print(f"  Consumo mensual estimado (suponiendo {DIAS_MES} días): {consumo_mensual:.2f} litros") # Formateado a 2 decimales.

print(f"\nConsumo en Metros Cúbicos:")
print(f"  Consumo diario estimado: {consumo_diario_m3:.4f} m³") # Formateado a 4 decimales.
print(f"  Consumo semanal estimado: {consumo_semanal_m3:.4f} m³") # Formateado a 4 decimales.
print(f"  Consumo mensual estimado: {consumo_mensual_m3:.4f} m³") # Formateado a 4 decimales.

print(f"\nEstimación de Costos (con tarifa de {costo_por_m3:.2f} por m³):") # Muestra la tarifa utilizada.
print(f"  Costo diario estimado: {costo_diario:.2f}") # Formateado a 2 decimales.
print(f"  Costo semanal estimado: {costo_semanal:.2f}") # Formateado a 2 decimales.
print(f"  Costo mensual estimado: {costo_mensual:.2f}") # Formateado a 2 decimales.

print(f"\nClasificación del consumo diario: {nivel_consumo}") # Muestra la clasificación obtenida.
```
